```python
import pandas as pd
import joblib

# 1. Carreguem la "memòria" que vam programar i desar a Barcelona
model_carregat = joblib.load('../models/model_vulnerabilitat_RF.joblib')
scaler_barcelona = joblib.load('../models/scaler_barcelona.joblib')

# 2. Carreguem les dades del nou municipi
print("📂 Carregant dades del nou municipi per avaluar...")
df_nou_poble = pd.read_csv('../data/raw/dades_girona.csv', sep=';')

# Seleccionem exactament les mateixes 5 variables de control
columnes_X = [
    'renda_disponible_Import_Euros', 
    'pad_dom_Llars_1_Avi_Sol', 
    'edificacions_any_Any_Mitja_Ponderat', 
    'heating_thermal_demand_intensity__2025-03-06',
    'torrid_nights__2024-01-01'
]

X_nou = df_nou_poble[columnes_X]

# 3. Apliquem l'escalador de Barcelona a les dades del nou poble
X_nou_scaled = scaler_barcelona.transform(X_nou)

# 4. Generem les prediccions de vulnerabilitat de forma immediata
print("🔮 Executant el motor predictiu de la Intel·ligència Artificial...")
df_nou_poble['PREDICIO_CVI'] = model_carregat.predict(X_nou_scaled)

# Guardem el resultat resultant
df_nou_poble.to_csv('../data/processed/vulnerabilitat_nou_municipi_predit.csv', index=False, sep=';')
print("✅ Càlcul finalitzat. Resultats cartogràfics preparats.")
```
